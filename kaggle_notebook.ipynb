# BoardMind: Whiteboard to Interactive Study Guide
**Track:** Multimodal Track | **Model:** Gemma 4 Vision

This notebook demonstrates the core inference pipeline used in the BoardMind application. We use a Gemma 4 model via the Google AI Studio API to ingest an image of a whiteboard and output a structured markdown study guide.

```python
import google.generativeai as genai
import PIL.Image
from google.colab import userdata

# In a real Kaggle environment, we would use Kaggle Secrets. 
# Here we simulate loading the API key.
API_KEY = userdata.get('GEMINI_API_KEY')
genai.configure(api_key=API_KEY)
```

```python
# Load a sample whiteboard image
img = PIL.Image.open('sample_whiteboard.png')
img.thumbnail((800, 800)) # Resize for faster inference
display(img)
```

```python
model = genai.GenerativeModel('gemma-4-31b-it')

prompt = """
You are an expert AI study assistant. 
Please analyze this whiteboard image thoroughly. 
Digitize all visible text, identify and describe any diagrams or flowcharts, 
extract any code or pseudocode, and produce a clean, well-structured 
Markdown study guide from the content.
"""

response = model.generate_content([prompt, img])
print(response.text)
```
